# Pandas Telemetry Processing

### Objective



- Pandas-based telemetry ingestion
- Time-indexing of telemetry data
- Query handling and filtering
- DataFrame creation and manipulation
- Timestamp conversion and management
- Time-range based telemetry selection
- Conditional data filtering
- Basic telemetry data inspection

```python
import pandas as pd

print("Pandas version:", pd.__version__)
```

```python
telemetry_data = {
    "timestamp": [
        "2026-08-10 09:00:00",
        "2026-08-10 09:01:00",
        "2026-08-10 09:02:00",
        "2026-08-10 09:03:00",
        "2026-08-10 09:04:00"
    ],
    "temperature": [25.4, 26.1, 27.3, 28.0, 27.6],
    "pressure": [1012, 1011, 1010, 1009, 1010],
    "altitude": [120, 125, 130, 135, 132],
    "speed": [45, 47, 50, 52, 49]
}

df = pd.DataFrame(telemetry_data)

df
```

```python
df.to_csv("telemetry_data.csv", index=False)

print("Telemetry data saved successfully.")
```

```python
telemetry_df = pd.read_csv("telemetry_data.csv")

telemetry_df
```

```python
print("Rows:", telemetry_df.shape[0])
print("Columns:", telemetry_df.shape[1])

print("\nColumn names:")
print(telemetry_df.columns.tolist())
```

```python
telemetry_df["timestamp"] = pd.to_datetime(telemetry_df["timestamp"])

print(telemetry_df["timestamp"])
```

```python
telemetry_df = telemetry_df.set_index("timestamp")

telemetry_df
```

```python
telemetry_df["timestamp"] = pd.to_datetime(telemetry_df["timestamp"])

telemetry_df = telemetry_df.set_index("timestamp")

telemetry_df
```

```python
# Time-Based Query Handling

time_range_data = telemetry_df.loc[
    "2026-08-10 09:01:00":"2026-08-10 09:03:00"
]

time_range_data
```

```python
print(telemetry_df.index)
print(telemetry_df.index.dtype)
print(type(telemetry_df.index))
```

```python
# Filtering telemetry data
filtered_data = telemetry_df[
    telemetry_df["temperature"] > 27
]

filtered_data
```

```python
# Sorting telemetry data by temperature
sorted_data = telemetry_df.sort_values(by="temperature", ascending=False)

sorted_data
```

```python
# Sorting telemetry data by temperature
sorted_data = telemetry_df.sort_values(by="temperature", ascending=False)

sorted_data
```

```python
# Statistical summary of telemetry data
summary = telemetry_df[["temperature", "pressure", "altitude", "speed"]].describe()

summary
```

```python
# Checking missing values
missing_values = telemetry_df.isnull().sum()

missing_values
```

```python
# Checking temperature range
temperature_check = telemetry_df[
    (telemetry_df["temperature"] < 20) |
    (telemetry_df["temperature"] > 40)
]

temperature_check
```

```python
# Checking pressure range
pressure_check = telemetry_df[
    (telemetry_df["pressure"] < 950) |
    (telemetry_df["pressure"] > 1010)
]

pressure_check
```

```python
# Checking altitude and speed ranges

altitude_check = telemetry_df[
    (telemetry_df["altitude"] < 0) |
    (telemetry_df["altitude"] > 500)
]

speed_check = telemetry_df[
    (telemetry_df["speed"] < 0) |
    (telemetry_df["speed"] > 200)
]

print("Altitude out-of-range records:")
print(altitude_check)

print("\nSpeed out-of-range records:")
print(speed_check)
```

```python
# Saving processed telemetry data

telemetry_df.to_csv("processed_telemetry.csv")

print("Processed telemetry data saved successfully.")
```

```python
# Final telemetry processing summary

print("Total records:", len(telemetry_df))
print("Columns:", list(telemetry_df.columns))
print("Data shape:", telemetry_df.shape)
print("Missing values:", telemetry_df.isnull().sum().sum())
print("Duplicate records:", telemetry_df.duplicated().sum())
```
