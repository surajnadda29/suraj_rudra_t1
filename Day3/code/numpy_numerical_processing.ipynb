```python
import numpy as np

print("RUDRAEDGE RE-T1 - Day 3")
print("NumPy Numerical Processing")
print("--------------------------")

print("NumPy Version:", np.__version__)
```

```python
# 1D sensor data array

temperature = np.array([24.5, 25.1, 26.3, 27.0, 26.7, 25.9])

print("Temperature Data:")
print(temperature)

print("\nArray Shape:", temperature.shape)
print("Array Size:", temperature.size)
print("Data Type:", temperature.dtype)
```

```python
# Accessing individual values using indexing

print("First Temperature:", temperature[0])
print("Third Temperature:", temperature[2])
print("Last Temperature:", temperature[-1])

# Slicing the array

print("\nFirst Three Readings:", temperature[:3])
print("Last Three Readings:", temperature[-3:])
```

```python
# NumPy array arithmetic operations

temperature_c = np.array([24.5, 25.1, 26.3, 27.0, 26.7, 25.9])

# Increase every temperature by 2°C
increased_temp = temperature_c + 2

# Convert Celsius to Fahrenheit
temperature_f = (temperature_c * 9/5) + 32

# Difference from a reference temperature
reference_temp = 25
temperature_difference = temperature_c - reference_temp

print("Original Temperature (°C):")
print(temperature_c)

print("\nTemperature after +2°C:")
print(increased_temp)

print("\nTemperature in Fahrenheit:")
print(temperature_f)

print("\nDifference from 25°C:")
print(temperature_difference)
```

```python
# Statistical operations on sensor data

print("Minimum Temperature:", np.min(temperature_c))
print("Maximum Temperature:", np.max(temperature_c))
print("Mean Temperature:", np.mean(temperature_c))
print("Median Temperature:", np.median(temperature_c))
print("Standard Deviation:", np.std(temperature_c))
```

```python
# Filtering sensor data using conditions

high_temperature = temperature_c[temperature_c > 26]

print("Temperatures above 26°C:")
print(high_temperature)

# Create a condition for normal temperature range
normal_temperature = temperature_c[
    (temperature_c >= 25) & (temperature_c <= 27)
]

print("\nTemperatures between 25°C and 27°C:")
print(normal_temperature)
```

```python
# 2D array representing multiple sensor readings

sensor_data = np.array([
    [24.5, 1012.3, 11.8],
    [25.1, 1011.8, 11.9],
    [26.3, 1011.2, 12.0],
    [27.0, 1010.7, 12.1],
    [26.7, 1010.9, 12.0]
])

print("Sensor Data:")
print(sensor_data)

print("\nShape:", sensor_data.shape)
print("Rows:", sensor_data.shape[0])
print("Columns:", sensor_data.shape[1])
```

```python
# Accessing data from a 2D NumPy array

print("First Row:")
print(sensor_data[0])

print("\nTemperature Column:")
print(sensor_data[:, 0])

print("\nPressure Column:")
print(sensor_data[:, 1])

print("\nVoltage Column:")
print(sensor_data[:, 2])
```

```python
# Vectorized numerical processing

# Apply calibration to all temperature readings
calibrated_temperature = temperature_c * 1.02 + 0.5

print("Original Temperature:")
print(temperature_c)

print("\nCalibrated Temperature:")
print(calibrated_temperature)
```

```python
# Comparing normal Python loop with NumPy vectorization

import time

data = np.random.random(1_000_000)

# Normal Python loop
start_loop = time.perf_counter()

loop_result = []
for value in data:
    loop_result.append(value * 2)

loop_time = time.perf_counter() - start_loop


# NumPy vectorized operation
start_vectorized = time.perf_counter()

vectorized_result = data * 2

vectorized_time = time.perf_counter() - start_vectorized


print("Loop Processing Time:", loop_time, "seconds")
print("Vectorized Processing Time:", vectorized_time, "seconds")

if vectorized_time > 0:
    print(
        "Speed Improvement:",
        round(loop_time / vectorized_time, 2),
        "x"
    )
```

```python
# Synthetic telemetry data using NumPy

np.random.seed(42)

samples = 1000

temperature = np.random.uniform(20, 35, samples)
pressure = np.random.uniform(990, 1030, samples)
voltage = np.random.uniform(11, 13, samples)
altitude = np.random.uniform(100, 500, samples)
speed = np.random.uniform(20, 80, samples)

print("Synthetic Telemetry Generated")
print("-----------------------------")
print("Number of Samples:", samples)

print("\nFirst 5 Temperature Values:")
print(temperature[:5])

print("\nFirst 5 Pressure Values:")
print(pressure[:5])

print("\nFirst 5 Voltage Values:")
print(voltage[:5])

print("\nFirst 5 Altitude Values:")
print(altitude[:5])

print("\nFirst 5 Speed Values:")
print(speed[:5])
```

```python
# Statistical analysis of synthetic telemetry

print("Telemetry Statistics")
print("--------------------")

print("Temperature Mean:", np.mean(temperature))
print("Temperature Min:", np.min(temperature))
print("Temperature Max:", np.max(temperature))

print("\nPressure Mean:", np.mean(pressure))
print("Pressure Min:", np.min(pressure))
print("Pressure Max:", np.max(pressure))

print("\nVoltage Mean:", np.mean(voltage))
print("Voltage Min:", np.min(voltage))
print("Voltage Max:", np.max(voltage))

print("\nAltitude Mean:", np.mean(altitude))
print("Altitude Min:", np.min(altitude))
print("Altitude Max:", np.max(altitude))

print("\nSpeed Mean:", np.mean(speed))
print("Speed Min:", np.min(speed))
print("Speed Max:", np.max(speed))
```

```python
# Telemetry validation using NumPy conditions

temperature_valid = (temperature >= 20) & (temperature <= 35)
pressure_valid = (pressure >= 990) & (pressure <= 1030)
voltage_valid = (voltage >= 11) & (voltage <= 13)
altitude_valid = (altitude >= 100) & (altitude <= 500)
speed_valid = (speed >= 20) & (speed <= 80)

print("Telemetry Validation")
print("--------------------")

print("Valid Temperature Readings:", np.sum(temperature_valid))
print("Valid Pressure Readings:", np.sum(pressure_valid))
print("Valid Voltage Readings:", np.sum(voltage_valid))
print("Valid Altitude Readings:", np.sum(altitude_valid))
print("Valid Speed Readings:", np.sum(speed_valid))

print("\nTotal Samples:", samples)
```

```python
# NumPy broadcasting

sensor_readings = np.array([
    [24.5, 1012.3, 11.8],
    [25.1, 1011.8, 11.9],
    [26.3, 1011.2, 12.0],
    [27.0, 1010.7, 12.1],
    [26.7, 1010.9, 12.0]
])

# Calibration offset for each sensor
calibration_offset = np.array([0.5, 2.0, 0.1])

calibrated_data = sensor_readings + calibration_offset

print("Original Sensor Data:")
print(sensor_readings)

print("\nCalibration Offset:")
print(calibration_offset)

print("\nCalibrated Sensor Data:")
print(calibrated_data)
```

```python
# Numerical processing routine

data = np.array([10, 20, 30, 40, 50], dtype=float)

addition_result = data + 5
multiplication_result = data * 2
square_result = data ** 2
sqrt_result = np.sqrt(data)

print("Original Data:")
print(data)

print("\nAfter Addition (+5):")
print(addition_result)

print("\nAfter Multiplication (×2):")
print(multiplication_result)

print("\nSquare of Data:")
print(square_result)

print("\nSquare Root of Data:")
print(sqrt_result)
```
